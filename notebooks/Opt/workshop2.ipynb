```python
import random
import matplotlib.pyplot as plt
import numpy as np
import numpy.linalg as la
```

```python
a, b, n = -1, 1, 200
u = np.linspace(a, b, n)
```

```python
def f1(x):
    return x**2

def f2(x):
    return x**2 + random.gauss(0, 0.5)

y1 = []
y2 = []

for i in u:
    y1.append(f1(i))
    y2.append(f2(i))
```

```python
z1 = np.poly1d(np.polyfit(u, y1, 3))

plt.plot(u, y1, '.', u, z1(u), '-')
plt.show()
z1
```
